```python
# 2: Imports e GPU

import json
import datetime
from pathlib import Path

import arviz as az
import numpy as np
import pandas as pd
import tensorflow as tf
import tensorflow_probability as tfp
import meridian
import pickle

from meridian import constants
from meridian.analysis import optimizer, summarizer, visualizer
from meridian.analysis.review import reviewer
from meridian.data import data_frame_input_data_builder
from meridian.model import model, prior_distribution, spec

import warnings
warnings.filterwarnings("ignore")

gpus = tf.config.list_physical_devices("GPU")
if gpus:
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)

```

```python
# 3: Dados

url = "https://raw.githubusercontent.com/google/meridian/main/meridian/data/simulated_data/csv/geo_all_channels.csv"
df = pd.read_csv(url)

if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

df['time'] = pd.to_datetime(df['time'], errors='coerce')

# Ajuste: Dummies e pop_geo
df['promo_dummy'] = (df['Promo'] > df['Promo'].mean()).astype(int)
df['pop_geo'] = np.random.randint(1000, 10000, len(df))  # Simula pop

print(f"✅ DataFrame carregado: {df.shape}")
```

```python
# 4: Builder InputData

from meridian.data import data_frame_input_data_builder

# Corrigido: nomes MAIÚSCULOS, iguais ao dataset
media_cols = [f"Channel{i}_impression" for i in range(5)]
media_spend_cols = [f"Channel{i}_spend" for i in range(5)]
media_channels = [f"Channel{i}" for i in range(5)]

control_cols = ["competitor_sales_control", "sentiment_score_control"]

input_data = (
    data_frame_input_data_builder.DataFrameInputDataBuilder(
        kpi_type='non_revenue',
        default_kpi_column='conversions',
        default_revenue_per_kpi_column='revenue_per_conversion'
    )
    .with_kpi(df)
    .with_controls(df, control_cols=control_cols)
    .with_media(df, media_cols, media_spend_cols, media_channels)
    .with_non_media_treatments(df, non_media_treatment_cols=["Promo"])
    .with_organic_media(
        df,
        organic_media_cols=["Organic_channel0_impression"],
        organic_media_channels=["organic_channel0"]
    )
    .with_population(df)
    .with_revenue_per_kpi(
        df,
        revenue_per_kpi_col="revenue_per_conversion",
        time_col="time"
    )
    .build()
)

print("✅ InputData criado!")
```

```python
set(df.columns)
```

```python
# 5: Roi

roi_mu, roi_sigma = 0.2, 0.9
build_media_channel_args = input_data.get_paid_media_channels_argument_builder()

roi_m = build_media_channel_args(
    Channel0=(0.15, 0.8),
    Channel1=(0.25, 1.0),
    Channel2=(0.2, 0.9),
    Channel3=(0.3, 0.7),
    Channel4=(0.18, 0.95)
)

roi_m_mu, roi_m_sigma = zip(*roi_m)

prior = prior_distribution.PriorDistribution(
    roi_m=tfp.distributions.LogNormal(roi_m_mu, roi_m_sigma, name=constants.ROI_M),
    mroi_m=tfp.distributions.LogNormal(0.0, 0.5),
    contribution_m=tfp.distributions.Beta(1.0, 99.0),
    beta_m=tfp.distributions.HalfNormal(5.0),
    gamma_c=tfp.distributions.Normal(0.0, 5.0),
    gamma_n=tfp.distributions.Normal(0.0, 5.0)
)
```

```python
# 6: Spec


# Número de períodos e geos
n_times_unique = len(df['time'].unique())
n_geos = df['geo'].nunique()

# Número de canais e controles
n_channels = len(media_channels)
n_controls = len(control_cols)
n_non_media = 1  # Promo

# ROI calibration (todos períodos calibrados)
roi_calibration_period = tf.constant(
    [[True] * n_channels] * n_times_unique,
    dtype=tf.bool
)

# Holdout: últimos 12 períodos
holdout_mask = [False] * max(0, n_times_unique - 12) + [True] * min(12, n_times_unique)
holdout_id = tf.constant([holdout_mask] * n_geos, dtype=tf.bool)

# Population scaling (todos controles)
control_population_scaling_id = tf.constant([True] * n_controls, dtype=tf.bool)
non_media_population_scaling_id = tf.constant([True] * n_non_media, dtype=tf.bool)

# Adstock customizado
adstock_decay_spec_custom = {
    'Channel0': 'geometric',
    'Channel1': 'geometric',
    'Channel2': 'binomial',
    'Channel3': 'binomial',
    'Channel4': 'geometric'
}

# --------------------------------------
# ModelSpec corretamente definido
# --------------------------------------

model_spec = spec.ModelSpec(
    prior=prior,
    media_effects_dist='log_normal',
    hill_before_adstock=False,
    max_lag=8,
    unique_sigma_for_each_geo=True,

    media_prior_type='roi',
    rf_prior_type='roi',
    paid_media_prior_type=None,

    roi_calibration_period=roi_calibration_period,
    rf_roi_calibration_period=None,

    organic_media_prior_type='contribution',
    organic_rf_prior_type='contribution',

    non_media_treatments_prior_type='coefficient',
    non_media_baseline_values=['min'],

    knots=None,
    baseline_geo=0,

    holdout_id=holdout_id,
    control_population_scaling_id=control_population_scaling_id,
    non_media_population_scaling_id=non_media_population_scaling_id,

    adstock_decay_spec=adstock_decay_spec_custom,

    enable_aks=True
)

print("ModelSpec criado com sucesso!")

```

```python
# 7: Sampling MCMC


mmm = model.Meridian(input_data=input_data, model_spec=model_spec)

import time
start_time = time.time()

# Amostrar o prior (opcional, mas recomendado)
mmm.sample_prior(500)

# Amostragem posterior MCMC
mmm.sample_posterior(
    n_chains=10,
    n_adapt=2000,
    n_burnin=500,
    n_keep=1000,
    seed=0
)

print(f"Tempo de execução: {time.time() - start_time} segundos")
```

```python
# 8: Teste Qualidade
reviewer.ModelReviewer(mmm).run()
```

```python
# 8.1: Diagnóstico - Posteriori
model_diagnostics = visualizer.ModelDiagnostics(mmm)
model_diagnostics.plot_rhat_boxplot()
```

```python
# 8.2: Fit
model_fit = visualizer.ModelFit(mmm)
model_fit.plot_model_fit()
```

```python
# 8.3: Resumo Modelo
mmm_summarizer = summarizer.Summarizer(mmm)
```
